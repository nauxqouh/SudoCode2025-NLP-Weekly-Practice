## 1. Compare output using API GPT and Gemini

```python
!pip install openai --quiet

from openai import OpenAI
client = OpenAI(api_key=OPENAI_API_KEY)

def call_gpt(prompt):
  response = client.responses.create(
      model="gpt-4o-mini",
      input=prompt
  )
  return response.output_text
```

```python
!pip install google-genai --quiet

from google import genai
client = genai.Client(api_key=GEMINI_API_KEY)

def call_gemini(prompt):
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt,
    )
    return response.text
```

```python
prompt = "Giải thích sự khác nhau giữa attention và self-attention trong 10 câu."
```

```python
print("\nGemini Output:")
print(call_gemini(prompt))
```

```python
print("\nGPT Output:")
print(call_gpt(prompt))
```

### Gemini vs GPT Comparison

| Criterion| Gemini| GPT |
|-------------------------|----------------------|---------------------------|
| **Explanation** | Very detailed, explains concepts like cross-attention, query/key/value, and self-attention in Transformers. | Concise, highlights key differences without excessive detail. |
| **Accuracy** | High accuracy for complex tasks due to detailed explanations. | Accurate for basic understanding and direct tasks. |
| **Speed**| Slower due to longer, verbose output. | Faster due to short and direct output. |
| **Cost**  | Higher, consumes more tokens.| Lower, consumes fewer tokens.|
| **Best Use Cases**| Deep learning explanations, complex QA, educational material. | Quick summarization, direct QA, large-scale text processing. |

**Note:**

- Gemini is better for educational purposes or in-depth QA. Gemini provides deeper insight.
- GPT is suitable for quick understanding. GPT is better for rapid responses or large-scale summarization. GPT is more cost-efficient for high-volume tasks.

### Result:

**Gemini:**

Gemini’s output is highly detailed and explanatory, making it well-suited for educational purposes or tasks that require deep understanding. It clearly distinguishes between attention and self-attention, explaining concepts like cross-attention, query/key/value vectors, and the role of self-attention in Transformers. This level of detail improves accuracy in complex QA tasks, helping the user grasp underlying mechanisms. However, because the output is longer and more verbose, it requires more tokens, which increases computational cost and can slow down processing in large-scale summarization or QA pipelines.

**GPT:**

GPT’s output is concise and direct, effectively capturing the key distinctions between attention and self-attention without excessive detail. This makes it ideal for quick summarization or straightforward QA tasks where speed and token efficiency are important. While it may not provide as deep an explanation as Gemini, it still maintains correct terminology and delivers accurate information. The shorter length reduces computational cost and allows for faster processing, making GPT outputs preferable for applications requiring rapid responses or large-scale text handling.

```python
import os
from openai import OpenAI

client = OpenAI(
    base_url="https://router.huggingface.co/v1",
    api_key=HF_TOKEN,
)

completion = client.chat.completions.create(
    model="meta-llama/Llama-3.2-3B-Instruct",
    messages=[
        {
            "role": "user",
            "content": "Explain the difference between attention and self-attention in 10 sentences."
        }
    ],
)

print(completion.choices[0].message.content)
```

## 3. LLM Context Caching

```python
long_prompt = "Đây là một đoạn văn bản rất dài được lặp lại nhiều lần. " * 200
long_prompt += "Mục tiêu là để đo lường hiệu suất của mô hình. Giờ đây, hãy tạo ra một danh sách gồm 10 mục về các loại GPU khác nhau. "

max_output_tokens = 100
```

```python
import time
from google import genai
from google.genai.errors import APIError

try:
    client = genai.Client(api_key=GEMINI_API_KEY)

    # 1. First Token Latency
    print(f"Started processing long prompt ({len(long_prompt.split())} words)...")

    start_time = time.time()

    # Stream mode
    response_stream = client.models.generate_content_stream(
        model='gemini-2.5-flash',
        contents=[long_prompt]
    )

    # Loop through stream to get the first token
    first_token_time = None
    output_tokens = 0
    total_latency = 0

    for chunk in response_stream:
        if not first_token_time:
            first_token_time = time.time()
            total_latency = first_token_time - start_time
            print("\nFirst Token !!")
            print(f"Pre-fill Time: {total_latency:.4f} seconds")
            print("Started Decoding - Generate token.")

        output_tokens += 1
        # Print for continuous progress
        print(chunk.text, end="", flush=True)

    end_time = time.time()

    # 2.Decoding Speed
    decoding_time = end_time - first_token_time
    subsequent_tokens = output_tokens - 1 if output_tokens > 0 else 0

    if subsequent_tokens > 0:
        avg_token_time = decoding_time / subsequent_tokens
        tokens_per_second = subsequent_tokens / decoding_time
        print(f"\n\nDecoding Result:")
        print(f"Total tokens: {output_tokens}")
        print(f"Average time to generate one token with KV Cache: {avg_token_time * 1000:.2f} ms")
        print(f"tokens/s: {tokens_per_second:.2f}")
    else:
        print("\nNot enough token to compute speed decoding.")

except APIError as e:
    print(f"\nError API: {e}. Please check your API.")
except Exception as e:
    print(f"\nERROR: {e}")
```
